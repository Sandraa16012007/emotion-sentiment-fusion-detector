## Load IMDb Dataset

Load the movie review dataset that will be used for sentiment classification.

```python
import pandas as pd

df = pd.read_csv(
    f"{PROJECT_ROOT}/data/imdb/imdb.csv"
)

df.head()
```

# Convert Sentiment Labels

Machine learning models require numerical labels.

```python
df["label"] = df["sentiment"].map({
    "negative": 0,
    "positive": 1
})

df.head()
```

```python
df.to_csv(
    f"{PROJECT_ROOT}/data/imdb/imdb_processed.csv",
    index=False
)

print("Processed dataset saved")
```
