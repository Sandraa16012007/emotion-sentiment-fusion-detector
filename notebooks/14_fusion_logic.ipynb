# Fusion Logic

Goal:

Combine:

1. Emotion classifier output
2. Sentiment classifier output

to detect alignment and mismatch.

Examples:

happy + positive → aligned

angry + positive → mismatch

sad + negative → aligned

happy + negative → mismatch

```python
emotion_to_group = {

    "happy": "positive",

    "angry": "negative",
    "sad": "negative",
    "fear": "negative",
    "disgust": "negative",

    "neutral": "neutral",

    "surprise": "neutral"
}
```

## Mapping Strategy

Positive:
- happy

Negative:
- angry
- sad
- fear
- disgust

Neutral:
- neutral
- surprise

Reason:

The fusion layer compares emotional polarity rather than exact emotion classes.

```python
def detect_mismatch(
    emotion_label,
    sentiment_label
):

    emotion_group = emotion_to_group[
        emotion_label
    ]

    if emotion_group == sentiment_label:
        return False

    if emotion_group == "neutral":
        return False

    return True
```

```python
print(
    detect_mismatch(
        "happy",
        "positive"
    )
)

print(
    detect_mismatch(
        "angry",
        "positive"
    )
)

print(
    detect_mismatch(
        "sad",
        "negative"
    )
)
```

```python
def fusion_confidence(
    emotion_confidence,
    sentiment_confidence
):

    return round(
        (
            emotion_confidence +
            sentiment_confidence
        ) / 2,
        3
    )
```

```python
fusion_confidence(
    0.82,
    0.91
)
```

```python
def build_output(
    emotion,
    sentiment,
    emotion_conf,
    sentiment_conf
):

    mismatch = detect_mismatch(
        emotion,
        sentiment
    )

    confidence = fusion_confidence(
        emotion_conf,
        sentiment_conf
    )

    return {

        "emotion": emotion,

        "sentiment": sentiment,

        "mismatch": mismatch,

        "emotion_confidence":
            emotion_conf,

        "sentiment_confidence":
            sentiment_conf,

        "fusion_confidence":
            confidence
    }
```

```python
result = build_output(

    emotion="angry",

    sentiment="positive",

    emotion_conf=0.88,

    sentiment_conf=0.92
)

result
```
